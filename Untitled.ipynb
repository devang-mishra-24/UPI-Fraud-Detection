```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
```

## Step 2 ~ Prepare and clean the data

```python
df = pd.read_csv("synthetic_upi_transactions_10000.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())
```

```python
# ------------------------------------------------------------
# 2. BASIC DATA QUALITY CHECK
# ------------------------------------------------------------

print("Dataset Shape:", df.shape)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# ------------------------------------------------------------
# 3. REMOVE DUPLICATES
# ------------------------------------------------------------

before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Duplicates Removed:", before - after)
print("Rows Remaining:", len(df))
```

```python
# ------------------------------------------------------------
# 4. HANDLE MISSING VALUES
# ------------------------------------------------------------

numeric_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns

# Numeric → median
for col in numeric_columns:
    df[col] = df[col].fillna(df[col].median())

# Categorical → mode
for col in categorical_columns:
    if df[col].mode().empty:
        df[col] = df[col].fillna("Unknown")
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# ------------------------------------------------------------
# 5. TIMESTAMP PROCESSING
# ------------------------------------------------------------

df["Timestamp"] = pd.to_datetime(
    df["Timestamp"],
    errors="coerce"
)

# Remove rows where timestamp could not be converted
df = df.dropna(subset=["Timestamp"])

print(df["Timestamp"].dtype)
```

```python
# ------------------------------------------------------------
# 6. TIME FEATURES
# ------------------------------------------------------------

df["hour"] = df["Timestamp"].dt.hour

df["day_of_week"] = df["Timestamp"].dt.dayofweek

df["day"] = df["Timestamp"].dt.day

df["month"] = df["Timestamp"].dt.month

df["is_weekend"] = (
    df["day_of_week"].isin([5, 6])
).astype(int)

# Night-time transaction
df["is_night"] = (
    df["hour"].isin([0, 1, 2, 3, 4, 5])
).astype(int)

display(
    df[
        [
            "Timestamp",
            "hour",
            "day_of_week",
            "is_weekend",
            "is_night"
        ]
    ].head()
)
```

```python
# ------------------------------------------------------------
# 7. SORT BY USER + TIME
# ------------------------------------------------------------

df = df.sort_values(
    by=["Sender ID", "Timestamp"]
).reset_index(drop=True)
```

```python
# ------------------------------------------------------------
# 8. USER AVERAGE TRANSACTION
# ------------------------------------------------------------

df["user_avg_amount"] = (
    df.groupby("Sender ID")["Amount"]
      .transform("mean")
)

df["user_median_amount"] = (
    df.groupby("Sender ID")["Amount"]
      .transform("median")
)
```

```python
# ------------------------------------------------------------
# 9. AMOUNT DEVIATION
# ------------------------------------------------------------

df["amount_deviation"] = (
    (df["Amount"] - df["user_avg_amount"])
    / (df["user_avg_amount"] + 1e-6)
)

df["amount_deviation_abs"] = (
    df["amount_deviation"].abs()
)
```

```python
# ------------------------------------------------------------
# 10. USER TRANSACTION FREQUENCY
# ------------------------------------------------------------

df["user_transaction_count"] = (
    df.groupby("Sender ID")["Transaction ID"]
      .transform("count")
)
```

```python
# ------------------------------------------------------------
# 11. TIME SINCE PREVIOUS TRANSACTION
# ------------------------------------------------------------

df["previous_transaction_time"] = (
    df.groupby("Sender ID")["Timestamp"]
      .shift(1)
)

df["time_since_last_transaction"] = (
    df["Timestamp"]
    - df["previous_transaction_time"]
).dt.total_seconds()

# First transaction of a user
df["time_since_last_transaction"] = (
    df["time_since_last_transaction"]
    .fillna(-1)
)
```

```python
# ------------------------------------------------------------
# 12. RECIPIENT REPETITION
# ------------------------------------------------------------

df["recipient_transaction_count"] = (
    df.groupby(
        ["Sender ID", "Receiver ID"]
    )["Transaction ID"]
    .transform("count")
)
```

```python
# ------------------------------------------------------------
# 13. DEVICE CHANGE
# ------------------------------------------------------------

df["previous_device"] = (
    df.groupby("Sender ID")["Device Type"]
      .shift(1)
)

df["device_changed"] = (
    (
        df["Device Type"] != df["previous_device"]
    )
    &
    df["previous_device"].notna()
).astype(int)
```

```python
# ------------------------------------------------------------
# 14. LOCATION CHANGE
# ------------------------------------------------------------

df["previous_location"] = (
    df.groupby("Sender ID")["Location"]
      .shift(1)
)

df["location_changed"] = (
    (
        df["Location"] != df["previous_location"]
    )
    &
    df["previous_location"].notna()
).astype(int)
```

```python
# ------------------------------------------------------------
# 15. USER'S USUAL TRANSACTION HOUR
# ------------------------------------------------------------

user_mean_hour = (
    df.groupby("Sender ID")["hour"]
      .transform("mean")
)

df["hour_deviation"] = (
    abs(df["hour"] - user_mean_hour)
)
```

```python
# ============================================================
# FIXED TRANSACTIONS IN LAST 1 HOUR
# ============================================================

df = df.sort_values(
    ["Sender ID", "Timestamp"]
).reset_index(drop=True)

df["transactions_last_1h"] = 0

for user, group in df.groupby("Sender ID", sort=False):

    times = group["Timestamp"]

    counts = (
        pd.Series(
            1,
            index=times
        )
        .rolling("1h", closed="left")
        .sum()
        .fillna(0)
        .values
    )

    df.loc[group.index, "transactions_last_1h"] = counts

df["transactions_last_1h"] = (
    df["transactions_last_1h"].astype(int)
)

print(df["transactions_last_1h"].describe())
```

```python
# ------------------------------------------------------------
# 17. USER SPENDING STATISTICS
# ------------------------------------------------------------

df["user_std_amount"] = (
    df.groupby("Sender ID")["Amount"]
      .transform("std")
)

df["user_std_amount"] = (
    df["user_std_amount"].fillna(0)
)

# Z-score-like behavioral feature
df["amount_zscore"] = (
    (df["Amount"] - df["user_avg_amount"])
    /
    (df["user_std_amount"] + 1e-6)
)
```

```python
# ------------------------------------------------------------
# 18. FEATURE CHECK
# ------------------------------------------------------------

print("New Shape:", df.shape)

display(df.head())

print("\nNew Features:")
print(df.columns.tolist())
```

```python
# ------------------------------------------------------------
# 19. SAVE PROCESSED DATA
# ------------------------------------------------------------

df.to_csv(
    "upi_transactions_processed.csv",
    index=False
)

print("Processed dataset saved successfully!")
```

## STEP 3 ~ IQR

```python
def iqr_bounds(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return Q1, Q3, IQR, lower_bound, upper_bound
```

```python
Q1, Q3, IQR, lower, upper = iqr_bounds(df["Amount"])

print("Amount IQR Analysis")
print("-------------------")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
df["iqr_amount_flag"] = (
    (df["Amount"] < lower) |
    (df["Amount"] > upper)
).astype(int)
```

```python
print(df["iqr_amount_flag"].value_counts())
```

```python
Q1, Q3, IQR, lower, upper = iqr_bounds(
    df["amount_deviation"]
)

print("Amount Deviation IQR Analysis")
print("-----------------------------")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
df["iqr_deviation_flag"] = (
    (df["amount_deviation"] < lower) |
    (df["amount_deviation"] > upper)
).astype(int)
```

```python
print(df["iqr_deviation_flag"].value_counts())
```

```python
Q1, Q3, IQR, lower, upper = iqr_bounds(
    df["transactions_last_1h"]
)

print("Transaction Frequency IQR Analysis")
print("-----------------------------------")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
df["iqr_frequency_flag"] = (
    (df["transactions_last_1h"] < lower) |
    (df["transactions_last_1h"] > upper)
).astype(int)
```

```python
Q1, Q3, IQR, lower, upper = iqr_bounds(
    df["amount_zscore"]
)

print("Amount Z-score IQR Analysis")
print("---------------------------")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
df["iqr_zscore_flag"] = (
    (df["amount_zscore"] < lower) |
    (df["amount_zscore"] > upper)
).astype(int)
```

```python
print(df["iqr_zscore_flag"].value_counts())
```

```python
iqr_flags = [
    "iqr_amount_flag",
    "iqr_deviation_flag",
    "iqr_frequency_flag",
    "iqr_zscore_flag"
]

df["iqr_score"] = df[iqr_flags].sum(axis=1)
```

```python
print("IQR Score Distribution")
print("----------------------")
print(df["iqr_score"].value_counts().sort_index())
```

```python
top_iqr = df.sort_values(
    ["iqr_score", "amount_deviation_abs"],
    ascending=False
)

display(
    top_iqr[
        [
            "Transaction ID",
            "Sender ID",
            "Timestamp",
            "Amount",
            "user_avg_amount",
            "amount_deviation",
            "amount_zscore",
            "transactions_last_1h",
            "iqr_amount_flag",
            "iqr_deviation_flag",
            "iqr_frequency_flag",
            "iqr_zscore_flag",
            "iqr_score"
        ]
    ].head(20)
)
```

## Step 4 ~ Isolation Forest

```python
features = [
    "Amount",
    "amount_deviation",
    "amount_zscore",
    "transactions_last_1h",
    "time_since_last_transaction",
    "recipient_transaction_count",
    "device_changed",
    "location_changed",
    "hour_deviation"
]

X = df[features].copy()

print("Feature shape:", X.shape)

print("\nFeatures used:")
print(features)
```

```python
print(X.isnull().sum())
```

```python
iso_forest = IsolationForest(
    n_estimators=200,
    contamination=0.10,
    random_state=42,
    n_jobs=-1
)
```

```python
iso_forest.fit(X)

print("Isolation Forest training completed!")
```

```python
df["isolation_prediction"] = iso_forest.predict(X)

print(df["isolation_prediction"].value_counts())
```

```python
df["isolation_flag"] = (
    df["isolation_prediction"] == -1
).astype(int)

print(df["isolation_flag"].value_counts())
```

```python
df["isolation_score"] = -iso_forest.score_samples(X)
```

```python
print(df["isolation_score"].describe())
```

```python
top_isolation = df.sort_values(
    "isolation_score",
    ascending=False
)

display(
    top_isolation[
        [
            "Transaction ID",
            "Sender ID",
            "Timestamp",
            "Amount",
            "amount_deviation",
            "amount_zscore",
            "transactions_last_1h",
            "device_changed",
            "location_changed",
            "hour_deviation",
            "isolation_score",
            "isolation_flag"
        ]
    ].head(20)
)
```

## STEP 6 ~ COMBINED FRAUD RISK ENGINE

```python
scaler = MinMaxScaler()

df["isolation_score_normalized"] = (
    scaler.fit_transform(
        df[["isolation_score"]]
    ).flatten()
)

print(
    df["isolation_score_normalized"].describe()
)
```

```python
df["iqr_score_normalized"] = (
    df["iqr_score"] / df["iqr_score"].max()
)

print(
    df["iqr_score_normalized"].describe()
)
```

```python
df["time_series_score_normalized"] = (
    df["time_series_score"] / df["time_series_score"].max()
)

print(
    df["time_series_score_normalized"].describe()
)
```

```python
df["fraud_risk_score"] = (
    0.40 * df["isolation_score_normalized"]
    + 0.35 * df["iqr_score_normalized"]
    + 0.25 * df["time_series_score_normalized"]
)
```

```python
def assign_risk_level(score):

    if score >= 0.75:
        return "CRITICAL"

    elif score >= 0.50:
        return "HIGH"

    elif score >= 0.25:
        return "MEDIUM"

    else:
        return "LOW"


df["risk_level"] = df["fraud_risk_score"].apply(
    assign_risk_level
)
```

```python
print(
    df["risk_level"].value_counts()
)
```

```python
top_risky_transactions = df.sort_values(
    "fraud_risk_score",
    ascending=False
)

display(
    top_risky_transactions[
        [
            "Transaction ID",
            "Sender ID",
            "Receiver ID",
            "Timestamp",
            "Amount",

            "iqr_score",
            "isolation_score",
            "time_series_score",

            "fraud_risk_score",
            "risk_level"
        ]
    ].head(25)
)
```

```python
df["fraud_flag"] = (
    df["risk_level"].isin(
        ["HIGH", "CRITICAL"]
    )
).astype(int)
```

```python
print(
    df["fraud_flag"].value_counts()
)
```

```python
def fraud_reason(row):

    reasons = []

    if row["iqr_score"] >= 2:
        reasons.append("Unusual transaction amount")

    if row["isolation_flag"] == 1:
        reasons.append("Multivariate anomaly")

    if row["rapid_transaction_flag"] == 1:
        reasons.append("Rapid transaction")

    if row["burst_flag"] == 1:
        reasons.append("Transaction burst")

    if row["unusual_hour_flag"] == 1:
        reasons.append("Unusual transaction time")

    if row["device_changed"] == 1:
        reasons.append("Device change")

    if row["location_changed"] == 1:
        reasons.append("Location change")

    if not reasons:
        return "Normal behavior"

    return ", ".join(reasons)


df["fraud_reason"] = df.apply(
    fraud_reason,
    axis=1
)
```

```python
display(
    df.sort_values(
        "fraud_risk_score",
        ascending=False
    )[
        [
            "Transaction ID",
            "Amount",
            "fraud_risk_score",
            "risk_level",
            "fraud_flag",
            "fraud_reason"
        ]
    ].head(20)
)
```
